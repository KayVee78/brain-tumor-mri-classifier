# **Part A - Application Area Review**

### **The Problem Domain**
**Domain**: Hospitals and Medicines<br/>
**Topic**: Automated Brain Tumor Classification Using Deep Learning on MRI Scans
<br/><br/>
Among the most complicated and life-threatening neurological disorders are brain tumours that can be of glioma and meningiomas types. The primary imaging modality in diagnosis is Magnetic Resonance Imaging (MRI) that is highly contrasting in soft tissues and images the brain structures in the absence of ionising radiation. However, the analysis of MRI images is an extremely specialised and time-consuming process that requires a high level of the skills of radiologists and neuro-oncologists.<br/>

Radiologists have to interpret hundreds of MRI slices per patient in various imaging sequences (including T1, T2 and FLAIR). This is a time-consuming, expensive and labour intensive manual process prone to human error of fatigue and inter-observer error. Research has investigated that there is a possibility of diagnostic errors especially in low-contrast tumours or at an early stage, which will defer treatment and adversely impact patient outcomes.<br/>

This has led to the increasing demand of automated systems that can assist the clinicians by providing reliable and fast tumour classification. Instead of replacing medical professionals, Artificial Intelligence (AI) is becoming more often placed as a **clinical decision-support or triage aid**, a second pair of eyes to minimize cases of missed diagnosis and more consistent diagnosis. The aim of automated brain tumour classification with the help of AI is to assist in early detection, prioritize the urgent cases, and enhance the efficacy of workflow in hospitals.<br/><br/>

### **Early Approaches**
Initial studies in automated brain tumour detection used classical machine learning algorithms, including Support Vector Machines (SVMs), k-Nearest Neighbours ( kNN), and Random Forests. These methods usually became a multi-stage pipeline that included image preprocessing, manual feature extraction and classification.<br/>

The Feature extraction was also a very crucial and difficult process in these systems. Domain experts have engineered features manually according to the tumour properties features, including shape, texture, intensity, and edge information. Such commonly used techniques were Gray-Level Co-occurrence Matrices (GLCM), histogram based features and wavelet transforms. These techniques were already partially successful but were extremely regulated by the quality and relevance of the handcrafted features.<br/>

The key limitation of the traditional approaches in machine learning were that they were not robust and generalisable. Tumours tend to be irregular in shape, heterogeneous in the texture and different among patients and imaging devices. Handcrafted features had difficulties in capturing this variability, which caused a reduced performance when models were used with unseen data. As a result of this, these systems could not be reliably effectively deployed in real world clinical practice.<br/><br/>

### **Deep Learning and CNNs**
The introduction of deep learning, especially Convolutional Neural Network (CNNs), led to a significant change in the field. CNNs are particularly created to handle grid-like information like pictures and can discover hierarchical representation of features directly by the values of raw pixels. As pointed out in medical imaging surveys, CNNs do not require any manual feature engineering and have demonstrated excellent results in a broad variety of imaging tasks (Litjens et al., 2017).<br/>

The CNNs are used to classify brain tumours and the low-level features (edges and textures) are learnt in the early layers of the network, and the higher layers attract more specific tumour patterns. According to the research by Pereira et al. (2016), in the task of classification of glioma, deep CNNs outperform classical approaches significantly, given that such studies are carried out on the basis of MRI data. In the same way, the progress in other architectures including VGG, ResNet and U-Net have enhanced accuracy and stability.<br/>

Large datasets that are annotated with deep learning models have also helped in research advancement and benchmarking like the Brain Tumor Segmentation (BraTS) challenge. The use of CNN based systems has reached a performance similar to, or even surpassing, human experts, especially in tumour detection and classification subtasks, in certain controlled tasks.<br/><br/>

### **Notable Results and Clinical Impact**
Many studies have found their classification accuracies to be over 90% on binary and multi-class brain tumour classification problems with CNN-based methodology. These findings support the possibility of AI systems to support radiologists and decrease their workload as well as increase the confidence of their diagnosis.<br/>

However, even though they are successful, deep learning systems have significant challenges. The large labelled datasets needed by CNNs are costly and time-intensive to develop in the medical field. Also, the interpretability of models is an issue to consider, and the adoption of the model in clinical practice must be transparent and trusting. Such methods as saliency maps or Grad-CAM have been suggested in order to enhance the explainability process by drawing attention to areas in the image that affect the model decision.<br/>

Notably, the recent studies focus on human-AI collaboration, in which AI systems are presented as a supporting tool, not a decision-maker. This complies with the standards of ethics and regulations in the healthcare industry, as well as with practical implementation conditions in hospitals.<br/><br/>

### **Summary**
AI-based brain tumour classification has come from more rigid and feature-based machine learning models, to more robust deep learning-based models that can learn directly on medical images. The CNN-based systems have proven to be robust and have a great potential in assisting the clinical, especially in the form of triage and decision-assistance devices. Although issues like data availability and interpretability are still present, these systems are constantly being improved and AI-assisted neuro-oncology is approaching the real-world application.

###Technique 1: Support Vector Machines (Classical Machine Learning)
One of the earliest machine learning algorithms that were used in the task of classifying brain tumours was the Support Vector Machines (SVMs). In this method, MRI images are pre-processed and transformed to numerical feature vectors with manual means of feature extraction (such as texture descriptors, histograms of intensity, or wavelet features). The features are then introduced into the SVM classifier to make the distinction between tumour and non-tumour classes using the optimum decision boundary.<br/><br/>

* Data availability:<br/>
SVMs do not need large datasets as deep learning models and are applicable in situations where there are limited labelled medical data. However, they are closely relying on the presence of well-designed handcrafted features.

* Time to setup:<br/>
The setup time is moderate. Feature engineering can be a labor intensive process and it can be linked to either trial-and-error or domain expertise. After features have been specified, SVM training is computationally cheap.

* Time to produce results:<br/>
The training and inference are quick, particularly when using tiny datasets. This leads SVMs to be ideal with fast experimentation but not as adaptive to variations of images of complex nature.

* Output:<br/>
The output is a class label (e.g., tumour or non-tumour) with a confidence margin. However, it has low interpretability and the model lacks spatial localisation of tumour regions.<br/><br/>

Strengths:

* Works well on small datasets

* Fast training and inference

* Good theoretical foundation.

Weaknesses:

* Manually extracted features.

* Poor generalisation to abnormal tumour forms.

* Poor scaling to high complexity imaging data.
<br/><br/>
As a result of these limitations, deep learning approaches have become the new replacement of SVMs in medical image devices (Pereira et al., 2016).


###Technique 2: Convolutional Neural Networks (Deep Learning)
The most popular method of automated brain tumour classification is the so-called Convolutional Neural Networks (CNNs). CNNs are input-autonomous operators of raw MRI images and acquire hierarchical features including edges, textures, and tumour-specific features by convolutional layers.<br/><br/>

* Data availability:<br/>
CNNs are very demanding in terms of large labelled datasets to be trained successfully. Their success has been achieved in large part because of public datasets like BraTS (Brain Tumor Segmentation), but the limited amount of data is still a problem in clinical settings.

* Time to setup:<br/>
The start-up cost is greater than in classical ML methods. The choice of model architecture, data augmentation, and the choice of hyperparameter should be carefully designed. However, recent models, like TensorFlow and PyTorch, make implementation much easier.

* Time to produce results:<br/>
The computation cost of training is high and time intensive, it usually involves the use of GPUs. After training, inference is rapid and can be used to provide clinical support in real-time or near-real-time.

* Output:<br/>
CNNs also give class probability and may be expanded to give a visual explanation with methods like Grad-CAM. This enhances clinical trust and interpretability.

Strengths:

* Automatic feature learning

* Good performance in classification.

* Robust to image variability

* The state of art performance in medical imaging.

Weaknesses:

* Needs huge labelled datasets.

* High computational cost

* Less transparency than simple models.
<br/><br/>
CNNs are the most appropriate option when it comes to real-world brain tumour classification systems because they have high performance (Litjens et al., 2017).


### Technique 3: Unsupervised Clustering (K-Means)
Unsupervised clustering algorithms like K-Means have also been experimented in brain MRI analysis, both in detection of tumour regions and exploration. In contrast to supervised techniques, clustering does not involve labelled data, and clusters pixels or image regions with similar ones.

* Data availability:<br/>
Clustering methods can also be used in the absence of labelled data and this is beneficial in medical fields where annotation is costly.

* Time to setup:<br/>
The setup time is also fairly low, with no labels being needed. However, preprocessing and choice of a desirable number of clusters are still difficult.

* Time to produce results:<br/>
Clustering algorithms are computationally inexpensive and the results are obtained faster hence are appropriate in preliminary analysis.

* Output:<br/>
The result is a representation of grouped image regions as opposed to the labeling of classes. Clusters have to be linked with tumour or healthy tissue with the help of clinical interpretation.

Strengths:

* No labelled data required

* Fast execution

* Can be used in exploratory analysis.

Weaknesses:

* Not suitable for reliable classification.

* Sensitive to the change in noise and intensity.

* Poor clinical applicability as a single solution.
<br/><br/>
As a result, clustering is normally treated as a supportive method and not a major way of classification.

### C(a) System Architecture Diagram

#### Problem Statement
This project aims to build an end-to-end deep learning pipeline that will be able to automatically classify brain MRI images into clinically relevant categories of tumors (e.g., glioma, meningioma, pituitary) and a no tumor category. The manual interpretation of MRI scans is prone to errors involving inconsistency accross readers, so te goal is to develop a system that takes MRI scans as input, consistent preprocessing, trains a CNN-based classifier and outputs reliable predictions with clear evaluation results on the unseen test data.

High level diagram.png

The system follows a 3 phase pipeline for brain tumour classification.
1) Phase 1 (Data Preparation): The brain MRI images are obtained via a kaggle dataset where four classes were used namely glioma, meningioma, pituitary, and no tumour. The data set is organised into directory-based training, validation and test splits, each divided into directories, and the images were resized to 150x150 and normalised to [0,1] range. The techniques of data augmentation of rotation, shifting, and flipping are only implemented during the training set to reduce overfitting.<br/>

* Data Ingestion – MRI images are loaded from the dataset directory (Training/ and Testing/).
* Label Mapping – Class labels are assigned based on folder names and a consistent label encoding is created.
* Preprocessing – All images are resized to 150 x 150 and RGB (3 channels) are kept for uniform input shape.
* Data Augmentation – ImageDataGenerator is used to apply transformations (e.g., rotation/shift/zoom) to improve generalization.
* Train/Validation Split – Training set is split into 80% training / 20% validation to tune and monitor learning without touching the test set.<br/><br/>

2) Phase 2 (Model development): The Convolutional Neural Network is used for feature extraction involving the stacking of Conv2D, ReLU and Maxpooling layers to learn spatial tumour features. Features extracted are fed through a classification head consisting of fully connected layers, dropout regularisation and Softmax output layer. The model is trained using the Adam optimiser with categorical cross-entropy loss over 15 epochs.

* Model Training – A CNN model is trained using the training generator, validating each epoch using the validation generator.
* Regularization & Control – Techniques like dropout are applied to reduce overfitting and stabilize training.<br/><br/>

3) Phase 3 (Evaluation and Testing) involves the evaluation of the performance of the model by measuring it in terms of accuracy curves, a confusion matrix, and a classification report. Lastly, the trained model would give predicted tumour classes and confidence scores based on the unseen MRI images.

* Final Testing – Final trained model is evaluated once on the separate Testing/ set to estimate real-world performance.
* Metrics & Reporting – Accuracy, precision/recall/F1, confusion matrix, and training curves are reported to justify performance.
* Deployment-Ready Output – Save the trained model and provide an inference step that predicts the class for a new MRI image.

```python
# Pre-processing & Data Loading
# Augmentation, Normalization, and Generator Creation

# Constants (Hyperparameters)
IMG_HEIGHT = 150
IMG_WIDTH = 150
BATCH_SIZE = 32
SEED = 42 # Ensures the 'random' shuffle is reproducible

# Define Data Generators
# Train Data Generator: Includes Augmentation + Rescaling + Validation Split
train_datagen = ImageDataGenerator(
    rescale=1./255,         # Normalize pixel values to 0-1
    rotation_range=15,      # Rotate slightly
    width_shift_range=0.1,  # Shift left/right
    height_shift_range=0.1, # Shift up/down
    horizontal_flip=True,   # Mirror image
    fill_mode='nearest',
    validation_split=0.2    # Reserve 20% for Validation
)

# Test Data Generator (Rescaling Only)
test_datagen = ImageDataGenerator(rescale=1./255)

# Create Generators (Load the Data)
print("--- Loading Training Data (80%) ---")
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training',
    seed=SEED,
    shuffle=True
)

print("\n--- Loading Validation Data (20%) ---")
val_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation',
    seed=SEED,
    shuffle=False
)

print("\n--- Loading Testing Data (Separate Folder) ---")
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

# Verification
class_names = list(train_generator.class_indices.keys())
print(f"\nDetected Classes: {class_names}")
```

**Data Handling and Dataset Separation**

* The dataset is handled using a clear and controlled separation strategy to prevent data leakage. **Training/** directory is further divided into **80% training** and **20% validation** train subsets with the fixed random seed to provide reproducibility.

* **Testing/** directory is loaded independently and is not utilized when training and validating the model. This ensures that validation metrics are only utilized in monitoring learning behaviour but final performance evaluation occurs only on unseen test data as shown in Section D.


```python
# Data Visualization & Statistics
# Exploratory Data Analysis (EDA) and Visual Inspection

# Quantitative Analysis (Class Counts)
# C(b) - Input Data Description
class_names = list(train_generator.class_indices.keys())

# Calculate counts per subset
train_counts = pd.Series(train_generator.classes).value_counts().sort_index()
val_counts = pd.Series(val_generator.classes).value_counts().sort_index()
test_counts = pd.Series(test_generator.classes).value_counts().sort_index()

# summary table
stats_df = pd.DataFrame({
    'Class Name': class_names,
    'Training Set': train_counts.values,
    'Validation Set': val_counts.values,
    'Testing Set': test_counts.values
})

print("--- Dataset Class Distribution ---")
display(stats_df)
print(f"Total Images: {stats_df.sum(numeric_only=True).sum()}")


# Visualisation Functions
def plot_class_representatives(generator):
    """Finding and plotting one clear example for each of the 4 classes"""
    images, labels = next(generator)

    plt.figure(figsize=(15, 4))
    found_classes = set()
    plot_idx = 1

    for i in range(len(images)):
        # Decode one-hot label
        label_idx = np.argmax(labels[i])
        class_name = class_names[label_idx]

        if class_name not in found_classes:
            plt.subplot(1, 4, plot_idx)
            plt.imshow(images[i])
            plt.title(f"Class: {class_name}")
            plt.axis('off')
            found_classes.add(class_name)
            plot_idx += 1

        if len(found_classes) == 4:
            break

    plt.suptitle("Representative Samples per Class", fontsize=14, y=1.05)
    plt.show()

def plot_augmented_grid(generator):
    """Plots a 3x3 grid to show Data Augmentation effects (Rotation/Shift)"""
    images, labels = next(generator) # Fetch a new batch of augmented data

    plt.figure(figsize=(8, 8))
    for i in range(9):
        plt.subplot(3, 3, i + 1)
        plt.imshow(images[i])
        label = class_names[np.argmax(labels[i])]
        plt.title(label)
        plt.axis('off')

    plt.suptitle("\n\n Augmented Training Samples (Randomly Rotated/Shifted)", fontsize=14, y=1.02)
    plt.show()

# Execute Visuals
print("\n--- Visualizing Input Data ---")
plot_class_representatives(train_generator)
plot_augmented_grid(train_generator)
```

**Technique Selection and Model Design**

* Convolutional Neural Network (CNN) was chosen to perform this task because CNNs are highly accurate in image-based problems because they are capable of capturing spatial hierarchies of features, including edges, textures, and shapes, that are essential in medical images.

* The model architecture comprises of three convolutional blocks, which provides the opportunity to extract low-level, mid-level, and high-level visual features in MRI scans in a gradual manner. The hierarchical structure allows the network to learn tumour-specific patterns, while also maintaining computational efficiency.

* A softmax activation function is used in the output layer to produce class probabilities for the four tumour categories. Training the model may be done by categorical cross-entropy loss, which suits in multi-class classification problems in which the labels are one-hot-encoded.


```python
# CNN Model Architecture
# Defining the Neural Network Structure (Feature Extraction + Classification)

# Initialize the Sequential Model
model = Sequential([
    # STAGE 1 FEATURE EXTRACTION
    # Input Layer: 150x150 RGB images
    Input(shape=(150, 150, 3)),

    # Convolution Block 1 Detects Low-level Features (Edges, Corners)
    Conv2D(filters=32, kernel_size=(3, 3), activation='relu'),
    MaxPooling2D(pool_size=(2, 2)), # Size reduced to 74x74

    # Convolution Block 2 Detects Mid-level Features (Shapes, Curves)
    Conv2D(filters=64, kernel_size=(3, 3), activation='relu'),
    MaxPooling2D(pool_size=(2, 2)), # Size reduced to 36x36

    # Convolution Block 3 Detects High-level Features (Tumor Textures)
    Conv2D(filters=128, kernel_size=(3, 3), activation='relu'),
    MaxPooling2D(pool_size=(2, 2)), # Size reduced to 17x17


    # STAGE 2 CLASSIFICATION
    Flatten(),

    Dense(512, activation='relu'),

    # Dropout: Randomly turns off 50% of neurons during training
    # This acts as "Regularization" to prevent memorizing the training data
    Dropout(0.5),

    # Output Layer: 4 Neurons (One for each tumor class)
    # Softmax converts raw numbers into probabilities
    Dense(4, activation='softmax')
])

# Compile the Model
# Adam: Adaptive Learning Rate
# Categorical Crossentropy: math formula for Multi-class errors
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Output Summary
model.summary()
```

```python
# Model Training
# Check if model exists (Save time) -> Train only if needed

# Configuration & Paths
USE_DRIVE = os.path.exists('/content/drive/MyDrive')
MODEL_PATH = '/content/drive/MyDrive/brain_tumor_classifier_best.keras' if USE_DRIVE else 'brain_tumor_classifier_best.keras'
HISTORY_PATH = '/content/drive/MyDrive/train_history.json' if USE_DRIVE else 'train_history.json'

print(f"Checking for model at: {MODEL_PATH}")

# Reset Generators & Calculate Steps Correctly (Resets to ensure it starts from the first image)
train_generator.reset()
val_generator.reset()

# Using len(generator) instead of manual division
steps_per_epoch = len(train_generator)
validation_steps = len(val_generator)

print(f"Training Steps: {steps_per_epoch} | Validation Steps: {validation_steps}")

# Skip Training if File Exists
if os.path.exists(MODEL_PATH):
# if False:
    print("\n✓ Found saved model! Loading from file...")
    print("   (Skipping training by using the pre-trained model)")
    model = load_model(MODEL_PATH)

    # Attempt to load history for the graphs
    history = None
    if os.path.exists(HISTORY_PATH):
        with open(HISTORY_PATH, 'r') as f:
            history_dict = json.load(f)
        print("✓ Training history loaded.")
    else:
        print("⚠︎ Model found, but history file missing. Graphs might be empty.")
        history_dict = {}

else:
    print("\n⚠︎ No saved model found. Starting Training (This may take time)...")

    # Define Callbacks (Stop early if not improving)
    callbacks = [
        EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True, verbose=1),
        ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=2, min_lr=1e-6, verbose=1),
        ModelCheckpoint(MODEL_PATH, monitor="val_loss", save_best_only=True, verbose=1)
    ]

    # Train
    history_obj = model.fit(
        train_generator,
        steps_per_epoch=steps_per_epoch,
        validation_data=val_generator,
        validation_steps=validation_steps,
        epochs=30,
        callbacks=callbacks,
        verbose=1
    )

    # Save History & Model reference
    history_dict = history_obj.history
    model.save(MODEL_PATH) # Save final version just in case

    # Save history to JSON for next time
    with open(HISTORY_PATH, "w") as f:
        json.dump(history_dict, f)

    print(f"\n✓ Training Complete. Best model saved to {MODEL_PATH}")

print("\nModel is ready for evaluation.")
```

```python
# Training History Visualization
# Plot Accuracy and Loss Curves

def plot_training_history(history_data):
    # Check if we have data to plot
    if not history_data:
        print("⚠︎ No training history found.")
        return

    # Extract Data
    acc = history_data.get('accuracy', [])
    val_acc = history_data.get('val_accuracy', [])
    loss = history_data.get('loss', [])
    val_loss = history_data.get('val_loss', [])

    epochs = range(1, len(acc) + 1)

    # Create Plot (Size 14x5 to match slide aspect ratio)
    plt.figure(figsize=(14, 5))

    # Plot 1: Accuracy
    plt.subplot(1, 2, 1)
    # 'bo-' = Blue Dots with Line (Training)
    plt.plot(epochs, acc, 'bo-', label='Training Accuracy')
    # 'r*-' = Red Stars with Line (Validation)
    plt.plot(epochs, val_acc, 'r*-', label='Validation Accuracy')

    plt.title('Training vs Validation Accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True)

    # Plot 2: Loss
    plt.subplot(1, 2, 2)
    plt.plot(epochs, loss, 'bo-', label='Training Loss')
    plt.plot(epochs, val_loss, 'r*-', label='Validation Loss')

    plt.title('Training vs Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)

    plt.show()

# Runs the function
plot_training_history(history_dict)
```

# **Part D - Testing and Evaluation**

### Testing Strategy and Verification of Correctness

* To ensure the trained CNN's correctness, this section confirms the presence of a held-out test dataset, which is then assessed against the trained model and the results of the model is compared to known ground-truth labels.

* Expected / correct results: <br/>
The ground-truth class labels of each of the MRI images are the Testing/ directory hierarchy (the true label is obtained by the folder the image is in: glioma, meningioma, pituitary, notumor).

* Produced results: <br/>
The results that are obtained are the predicted class values that the trained CNN model produces, produced by the softmax output probabilities on the model (the most likely class will be the one with the highest probability).<br/><br/>

**Testing Methods Used**

* Held-out test set evaluation:<br/>
Testing/ dataset is evaluated only, and it is stored apart and not applied either in training or validation to allow an impartial estimation of generalisation.

* Prediction alignment: Deterministic: <br/>
The test data generator is set with `shuffle=False` to ensure that the sequence of model predictions is the same as the sequence of ground-truth labels so that one can make the expected vs produced comparison correctly.

* Measurements of quantitative evaluation: <br/>
The measure of correctness is based on a confusion matrix and classification report (precision, recall, F1-score, and support per class) that present evidence of a good performance both in total and per class.

* Qualitative inspection: <br/>
The test MRI images are visually examined by visualizing the image and the true label, predicted label and confidence score (percentage) of the model which gives an interpretable gauge on model behaviour as opposed to aggregate measures.

### Test Set Evaluation: Expected vs Produced Results

```python
# Model Evaluation on Test Set
# Generate Confusion Matrix and Classification Report (Precision/Recall)

# Generate Predictions
test_generator.reset()

print("Generating predictions on the Test Set...")
# verbose=1 shows a progress bar
predictions = model.predict(test_generator, verbose=1)

predicted_classes = np.argmax(predictions, axis=1)

# Get the True labels from the generator
true_classes = test_generator.classes
class_labels = list(test_generator.class_indices.keys())

# Confusion Matrix
cm = confusion_matrix(true_classes, predicted_classes)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_labels,
            yticklabels=class_labels)
plt.title('Confusion Matrix', fontsize=16)
plt.ylabel('True Label', fontsize=12)
plt.xlabel('Predicted Label', fontsize=12)
plt.show()

# Classification Report & Accuracy
print("\n--- Detailed Classification Report ---")
# calculates Precision, Recall, and F1-Score for each class
report = classification_report(true_classes, predicted_classes, target_names=class_labels)
print(report)

# Overall Accuracy
final_acc = accuracy_score(true_classes, predicted_classes)
print(f"\n✓ Overall Test Accuracy: {final_acc*100:.2f}%")
```

In addition to aggregate performance metrics, further correction testing was done by applying the complementary testing practices. To ensure that the CNN is significantly better than chance-level predictions and predictions driven by imbalances, a **baseline comparison** was done using some strategies (majority-class prediction and random guessing). Another **sanity test** was done to overfit the model on a very small training subset to ensure that the learning pipeline (pre-processing, label alignment, loss computation and optimisation) was set up correctly. Lastly, a few **individual test cases** were checked by making a direct comparison between the predicted labels of the expected classes and the results of the predictions and the confidence scores, and a direct comparison with human-checkable evidence of correctness at the sample level.




**Baseline Comparison**

```python
# ==== Baseline Comparison ====

# Computed final test accuracy
cnn_test_acc = np.mean(predicted_classes == true_classes)

# Majority-class baseline (always predict the most frequent true class)
counts = np.bincount(true_classes)
majority_class_id = np.argmax(counts)
majority_baseline_acc = counts[majority_class_id] / len(true_classes)

# Random baseline (uniform guessing over K classes)
K = len(class_labels)
random_baseline_acc = 1.0 / K

print("\n--- Baseline Comparison (Correctness Evidence) ---")
print("Test set class counts:", dict(zip(class_labels, counts)))
print(f"Majority class: {class_labels[majority_class_id]}")
print(f"Majority baseline accuracy: {majority_baseline_acc*100:.2f}%")
print(f"Random baseline accuracy (1/{K}): {random_baseline_acc*100:.2f}%")
print(f"\nCNN test accuracy: {cnn_test_acc*100:.2f}%")
print(f"Improvement over majority baseline: +{(cnn_test_acc - majority_baseline_acc)*100:.2f}%")
print(f"Improvement over random baseline:   +{(cnn_test_acc - random_baseline_acc)*100:.2f}%")
```

Both the majority-class and random baselines has lower accuracy compared to the CNN which justifies the fact that the model is being trained on discriminative features and not on either class imbalance or chance-level guessing.

**Sanity Test**

```python
# ==== Sanity Test - Overfit a tiny subset (one mini-batch) ====

train_generator.reset()
x_small, y_small = next(train_generator)

# Clone model architecture so the trained model isn't affected
sanity_model = tf.keras.models.clone_model(model)
sanity_model.build(input_shape=(None, x_small.shape[1], x_small.shape[2], x_small.shape[3]))

sanity_model.compile(
    optimizer=Adam(learning_rate=1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("\n--- Sanity Test: Overfitting one mini-batch ---")
history_sanity = sanity_model.fit(
    x_small, y_small,
    epochs=25,
    verbose=0
)

sanity_loss, sanity_acc = sanity_model.evaluate(x_small, y_small, verbose=0)
print(f"Sanity subset accuracy: {sanity_acc*100:.2f}%")
print("Last 5 training accuracies:", [round(v*100, 2) for v in history_sanity.history['accuracy'][-5:]])
```

Getting extremely high accuracy on one mini-batch means that the training pipeline (labels, preprocessing, model compilation, and optimisation) is operating properly. In case the model was not able to overfit this small subsample, it would indicate a pipeline problem like label discrepancy or faulty input processing.

**Individual Test Cases**

```python
# ==== Individual Test Cases: Expected vs Produced (with confidence) ====

np.random.seed(42)
n_cases = 10

# Choose random test indices
idxs = np.random.choice(len(true_classes), size=n_cases, replace=False)

rows = []
for i in idxs:
    true_id = int(true_classes[i])
    pred_id = int(predicted_classes[i])
    conf = float(np.max(predictions[i]))  # predictions must be probability vector per sample

    rows.append({
        "Test image path": test_generator.filepaths[i],
        "Expected (true)": class_labels[true_id],
        "Produced (pred)": class_labels[pred_id],
        "Confidence": f"{conf*100:.2f}%",
        "Match?": "✓" if true_id == pred_id else "✗"
    })

df_cases = pd.DataFrame(rows)
df_cases
```

### Prediction Demonstration (Qualitative Evidence)

```python
# Prediction Demonstration
# Displays test images with predicted vs actual labels

def visualize_predictions(generator, model):
    # Get a batch of test data
    generator.reset()
    x_test, y_test = next(generator)

    # Get predictions for this batch
    preds = model.predict(x_test, verbose=0)

    # Setup the Grid (2 rows x 4 columns = 8 images)
    plt.figure(figsize=(16, 8))

    # loop through the first 8 images of the batch
    for i in range(8):
        plt.subplot(2, 4, i + 1)

        plt.imshow(x_test[i])

        true_index = np.argmax(y_test[i])
        pred_index = np.argmax(preds[i])
        confidence = preds[i][pred_index] * 100

        class_map = {v: k for k, v in generator.class_indices.items()}
        true_label = class_map[true_index]
        pred_label = class_map[pred_index]

        color = 'green' if true_index == pred_index else 'red'

        plt.title(f"True: {true_label}\nPred: {pred_label} ({confidence:.2f}%)",
                  color=color, fontsize=11, fontweight='bold')
        plt.axis('off')

    plt.suptitle("Model Predictions on Unseen Test Data", fontsize=16)
    plt.tight_layout()
    plt.show()

# Run the visualizer
# Note: since test_generator is shuffle=False (standard for Confusion Matrix), this will always show the first 8 images.
visualize_predictions(test_generator, model)
```

#Part E - **Equity, Diversity, Inclusion (EDI) and Sustainability**

In this section, the proposed AI-based Brain Tumour Classification System is critically assessed both in the view of the society and the environment. Since the system will be used in healthcare, the responsible development of AI should be conducted to guarantee the equitable, accessible, ethical, and sustainable implementation.
<br/><br/>

## 1) Equity, Diversity and Inclusion (EDI).

**1.1 Bias and Fairness**

Medical AI systems, especially those trained on small or non-representative datasets, are rather prone to bias. The brain MRI data that trains deep learning models are usually limited to a certain geographic area, hospital or even a demographic group. Consequently, the trained model can work well with one group of population and not be good with another, e.g., patients from different ethnic backgrounds, age, and vendors of imaging equipment.

When dealing with brain tumour classification, biased prediction may result in inequality of diagnostic performance that may be damaging to underrepresented groups of patients. In order to reduce this risk, one should consider diverse datasets, conduct stratified assessment across the demographic sub groups when feasible, and continuously monitor model performance after deployment. The given system is thus placed as a decision-support tool, which does not imply that the end clinical decision will be left to the qualified medical professionals. <br/><br/>

**1.2 Accessibility and Inclusion**

The healthcare AI systems should not only be available in advanced hospitals but also be affordable to the resource-limited clinical environments. The CNN-based systems often need specialised hardware like GPUs and this might not be available in smaller hospitals or developing areas. This poses a threat of increasing the digital divide in accessing healthcare.

In order to further improve inclusivity, it is possible to deploy the system with centralised inference servers, with the hospitals uploading scans and receiving predictions without ever having to maintain the costly local infrastructure. Also, the output of the system must be delivered in an understandable format that is easy to interpret like probability scores and visual explanations to accommodate clinicians with different levels of AI familiarity.

The system is intended to help the radiologists and not to substitute them so that the existing clinical processes are not neglected because human expertise is still needed.<br/><br/>

**1.3 Ethical Implications**

The issue of AI use in medical diagnosing has far-reaching ethical concerns. False assumptions may cause delays in treatment or the needless worry of the patients. Hence, accountability and transparency are paramount. This system has to be explicit in its limitations and confidence levels, and the clinicians ought to be taught how to interpret AI outputs in a responsible manner.

Another important ethical issue is patient privacy. The data provided by MRI is very sensitive, and it has to be ensured that data protection rules are observed such as data anonymisation, safe storage, and adherence to healthcare data rules. Informed consent is also necessary when the AI-related analysis is applied in clinical decision-making as a form of ethical implementation.

Generally, there is ethical responsibility as there is human oversight, transparency, and strict data governance.<br/><br/>

## 2) Environmental Sustainability.

**2.1 Computational Resource Analysis**

Deep learning models especially CNNs demand a lot of computational resources to train. To train these models, one generally does repeated forward and backward passes of large image sets, which in many cases require energy-consuming GPUs. This leads to a significant carbon footprint when models are retrained on a regular basis.

However, CNNs are quite efficient in inference after they are trained. In the clinical domain, inference is significantly more common than training, and hence the total environmental cost can be decreased by cutting down training loops and reusing already trained models.<br/><br/>

**2.2 Data Storage Requirements**

Medical imaging datasets are storage-intensive since MRI scans are of high resolution. Raw image, augmented data and model checkpoints storage can pose a significant energy-use threat to data centres especially when performed over a long period.

To minimize effects of storage, unnecessary duplicates can be eliminated, and compress image formats can be applied where clinically acceptable. Essential model versions can be stored and the old experiments can be archived down further reducing storage needs.<br/><br/>

**2.3 Reducing Environmental Impact in Production**

To enhance sustainability in production deployment, below mentioned strategies can be used:

* Transfer learning may be applied to save training time and energy by training pretrained models rather than performing training.

* Pruning and quantisation are the model optimisation methods that can be used to minimise the size and the cost of the model.

* Cloud-based deployment with energy-efficient data centres can lower the carbon emissions compared to on-premise infrastructure.

* Scheduled retraining only when performance degradation is detected leads to avoid unnecessary computational expenditure.

With these strategies, the system will create the balance between high diagnostic performance and responsible environmental impact.<br/><br/>
